```python
import requests
import json
from datetime import datetime

#start_date="2026-08-01"
#end_date="2026-09-10"

url=f'https://earthquake.usgs.gov/fdsnws/event/1/query?format=geojson&starttime={start_date}&endtime={end_date}'

try:
    response=requests.get(url)

    if response.status_code==200:
        print(f'Connection OK')
    else: 
        print(f"Request failed: HTTP {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f'Request error: {e}')

if response.status_code==200:
    data=response.json()
    data=data['features']

    path= f'/lakehouse/default/Files/{start_date}_earthquake_data.json'

    with open(path,'w') as file:
        json.dump(data, file, indent=4)

    print(f"Data is successfully loaded into {path}")

else:
    print(f"Fail to fetch data, status code:", response.status_code)

print("========== BRONZE INGESTION ==========")
print(f"Run time: {datetime.now()}")
print(f"Start date: {start_date}")
print(f"End date: {end_date}")
print(f"HTTP status: {response.status_code}")
print(f"Records received: {len(data)}")
```
